# Basic concepts of programming

ပရိုဂရမ်ရေးခြင်းဆိုသည်မှာ computer ၏ **input** နှင့် **output** ကို အစီအစဉ်တကျ ထိန်းချုပ်ခြင်း ဖြစ်ပါသည်။


# sep

**print function** သည် တန်ဖိုးများကို ပြသရာတွင် တန်ဖိုးနှစ်ခုကြားတွင် space တစ်ခုခြား၍ ပြသပေးပါသည်။

အကယ်၍ space နေရာတွင် တစ်ခြားစာတန်ဖိုးများ ပြသလိုပါက **sep** ဟူသော keyword ကို အသုံးပြုနိုင်ပါသည်။

ဥပမာအားဖြင့် တန်ဖိုးနှစ်ခုကြားတွင် **ပုဒ်ကလေး** နှင့် **‌space** ခြားပြီး ပြသလိုပါက အောက်ပါအတိုင်း ရေးသားနိုင်ပါသည်။


```python
x = "Apple"
y = "Banana"
z = "Orange"
a = "Mango"
print(x, y, z, a, sep="၊ ")
```

စာကြောင်း၏အဆုံးတွင် ပေါင်းပေးမည့် စာတန်ဖိုးဖြစ်ပါသည်။

default အနေဖြင့် new line character ဖြစ်သော **"\n"** ကို စာကြောင်း၏အဆုံးတွင် ထပ်ပေါင်းပေးပါသည်။

new line character သည် မမြင်ရသော စာအမျိုးအစားဖြစ်ပြီး စာကြောင်းတစ်ခုကို အဆုံးသတ်ပေးသော လုပ်ဆောင်ချက်ကို လုပ်ဆောင်ပါသည်။

လက်ရှိစာကြောင်းကို အဆုံးသတ်ပေးပြီး စာကြောင်းအသစ်ကို စတင်ပေးသဖြင့် **line feed** ဟူ၍လည်း ခေါ်ကြပါသေးသည်။


ထို့ကြောင့် print() တစ်ကြိမ် အသုံးပြုတိုင်းတွင် စာအသစ်တစ်ကြောင်း ဖြစ်ပေါ်ပါသည်။





```python
x = "Apple"
y = "Banana"
z = "Orange"
a = "Mango"

print(x, y, z, a)
print(z, y, x)
print(a)
```
